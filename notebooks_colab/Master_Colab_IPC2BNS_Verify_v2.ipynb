## ⚙️ Step 0: Google Drive Mount & Environment Initialization
This cell automatically connects to your Google Drive folder (`NLP-rspaper`), sets the working directory, and installs all requirements.

```python
import os, sys
from pathlib import Path

try:
    import google.colab
    IN_COLAB = True
    print("[Colab] Detected Google Colab. Mounting Google Drive...")
    from google.colab import drive
    drive.mount('/content/drive', force_remount=False)
    
    # Check Google Drive project directory paths
    drive_candidates = [
        Path('/content/drive/MyDrive/NLP-rspaper'),
        Path('/content/drive/MyDrive/NLP_rs'),
        Path('/content/drive/MyDrive/research paper/NLP_rs'),
        Path.cwd()
    ]
    
    for cand in drive_candidates:
        if (cand / 'code' / 'src').exists():
            os.chdir(cand)
            print(f"[Colab] Working directory set to: {os.getcwd()}")
            break
            
    if Path("requirements.txt").exists():
        print("[Colab] Installing dependencies from requirements.txt...")
        get_ipython().system("pip install -q -r requirements.txt")
except ImportError:
    IN_COLAB = False
    print("[Local] Running in local environment:", os.getcwd())

# Ensure project root is on sys.path
root_candidates = [Path.cwd(), Path.cwd().parent, Path.cwd() / 'code']
for p in root_candidates:
    if (p / 'code' / 'src').exists():
        sys.path.insert(0, str(p / 'code'))
        break
    elif (p / 'src').exists():
        sys.path.insert(0, str(p))
        break

print("[Setup Complete] Current Directory:", os.getcwd())
print("[Setup Complete] Code Root on sys.path:", sys.path[0])
```

## ⏳ Step 1: Phase 1 — Temporal & Savings Clause Engine
Evaluates **Article 20(1)** non-retroactivity, **Section 531(2)(a) BNSS** savings, and **Section 358 BNS** on transitional scenarios.

```python
from src.temporal.timeline_parser import TimelineParser
from src.temporal.savings_clause_engine import SavingsClauseEngine

parser = TimelineParser()
engine = SavingsClauseEngine()

sample_query = "Assault occurred on 10 June 2024. FIR registered on 15 July 2024."
timeline = parser.parse(sample_query)
res = engine.resolve_timeline(timeline)

print("Query:", sample_query)
print("Substantive Law:", res.substantive_code, "(Art 20(1) Mandate)")
print("Procedural Law :", res.procedural_code, "(BNSS 2023 - No Pending FIR prior to July 1)")
print("Rationale      :", res.reasoning)
```

## 🛡️ Step 2: Phase 2 — Discrete Multi-Stage Gate Verifiers
Tests the 4-stage mutation verifier (Timeline paradox, Regime leakage, Concordance veto, Grounding invariant).

```python
from src.verifier.stage_leakage_verifier import StageLeakageVerifier

verifier = StageLeakageVerifier()
print("Testing Stage 1 (Timeline Paradox):", verifier.verify_stage1_timeline({"incident_date": "2024-08-01", "fir_date": "2024-05-01"}))
print("Testing Stage 3 (Repealed Citation):", verifier.verify_stage3_citations(["IPC §124A"], is_post_july_offence=True))
```

## 🏛️ Step 3: Phase 3 — Query Router & Model Council
Routes queries across Tier 1 (~5ms), Tier 2 (~45ms), and Tier 3 (~150ms), utilizing disk checkpoints.

```python
from src.council.model_council import ModelCouncil

council = ModelCouncil()
verdict = council.process_query("What is the section for cheating and dishonestly inducing delivery in BNS?")
print("Tier:", verdict.routing_tier)
print("Consensus:", verdict.consensus_reached, "| Confidence:", verdict.consensus_confidence)
print("Primary Citations:", verdict.primary_citations)
print("From Cache:", verdict.from_cache)
```

## 🛑 Step 4: Phase 4 — Selective Prediction & Conflict Cards
